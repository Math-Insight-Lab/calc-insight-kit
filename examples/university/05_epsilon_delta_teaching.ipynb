## 1. 严格定义回顾

$$\lim_{x \to x_0} f(x) = L \iff \forall \varepsilon > 0, \exists \delta > 0, \text{ s.t. } 0 < |x - x_0| < \delta \implies |f(x) - L| < \varepsilon$$

## 2. 数值实验：寻找最佳 $\delta$ 函数

对于 $f(x) = \frac{\sin x}{x}$，$x_0 = 0$，$L = 1$

```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt
from calc_insight_kit.config import FIG_SIZE, DPI

# 数值实验：给定不同 \u03b5，寻找最大 \u03b4
def find_max_delta(eps, x0=0, L=1.0,
                   f_func=lambda t: np.sin(t)/t if t != 0 else 1.0,
                   x_range=(-2, 2)):
    xs = np.linspace(x_range[0], x_range[1], 10000)
    valid_mask = (np.abs(f_func(xs) - L) < eps) & (xs != x0)
    valid_xs = xs[valid_mask]
    if len(valid_xs) == 0:
        return 0
    delta_left = np.max(x0 - valid_xs[valid_xs < x0]) if np.any(valid_xs < x0) else 0
    delta_right = np.max(valid_xs[valid_xs > x0] - x0) if np.any(valid_xs > x0) else 0
    return min(delta_left, delta_right)

epsilons = [0.5, 0.3, 0.1, 0.05, 0.01]
print('\u03b5\u00a0\u00a0| \u00a0\u00a0\u03b4\u00a0\u00a0\u00a0| \u00a0\u00a0\u03b4/\u03b5')
print('-' * 30)
for eps in epsilons:
    d = find_max_delta(eps)
    print(f'{eps:5.2f} | {d:8.4f} | {d/eps:8.2f}')
```

## 3. 误差区间多参数可视化

```python
x = np.linspace(-2, 2, 1000)
y = np.sin(x) / x
x0, L = 0, 1.0

fig, axes = plt.subplots(2, 3, figsize=(14, 8))
axes = axes.flatten()

for idx, eps in enumerate([0.5, 0.3, 0.1, 0.05, 0.02, 0.01]):
    d = find_max_delta(eps)
    ax = axes[idx]
    ax.plot(x, y, color='#2c3e50', linewidth=1.5)
    ax.axhline(L + eps, c='#e74c3c', ls='--', alpha=0.5)
    ax.axhline(L - eps, c='#e74c3c', ls='--', alpha=0.5)
    ax.axvline(x0 + d, c='#27ae60', ls='--', alpha=0.5)
    ax.axvline(x0 - d, c='#27ae60', ls='--', alpha=0.5)
    ax.scatter(x0, L, c='orange', s=80, zorder=10)
    ax.set_title(f'\u03b5={eps:.2f}, \u03b4={d:.4f}')
    ax.grid(alpha=0.3)

plt.tight_layout()
plt.savefig('university_epsilon_delta_analysis.png', dpi=150)
plt.show()
```

## 4. 严格证明练习

**定理：** 设 $f(x) = \frac{x^2-1}{x-1}$，证明 $\lim_{x \to 1} f(x) = 2$。

**证明：**
1. 化简：$f(x) = x + 1$（当 $x \neq 1$）
2. 给定 $\varepsilon > 0$，需找到 $\delta > 0$ 使得：
   $$0 < |x - 1| < \delta \implies |f(x) - 2| < \varepsilon$$
3. 由于 $|f(x) - 2| = |x + 1 - 2| = |x - 1|$，直接取 $\delta = \varepsilon$
4. 验证完成：$0 < |x - 1| < \varepsilon \implies |f(x) - 2| < \varepsilon$ ✓

**思考题：** 对于 $g(x) = \frac{\sin x}{x}$，证明 $\lim_{x \to 0} g(x) = 1$。

```python
# 泰勒展开验证
x = sp.Symbol('x')
taylor = (sp.sin(x)/x).series(x, 0, n=7).removeO()
print(f'sin(x)/x 的麦克劳林展开：{taylor}')
print(f'当 x -> 0 时，极限值为：{taylor.subs(x, 0)}')
```
