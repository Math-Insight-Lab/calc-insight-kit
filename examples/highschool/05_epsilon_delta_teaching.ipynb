## 1. ε-δ 定义的直观理解

极限的定义：当 $x$ 在 $x_0$ 的 $\delta$ 邻域内时，$f(x)$ 在极限值 $L$ 的 $\varepsilon$ 邻域内。

数学表达：$$\forall \varepsilon > 0, \exists \delta > 0, \text{ s.t. } 0 < |x - x_0| < \delta \implies |f(x) - L| < \varepsilon$$

## 2. 例题分析

**例题：** 证明 $\lim_{x \to 0} \frac{\sin x}{x} = 1$

步骤：
1. 确定 $x_0 = 0$，$L = 1$
2. 给定任意 $\varepsilon > 0$，找到对应的 $\delta > 0$
3. 验证：当 $0 < |x| < \delta$ 时，$|\frac{\sin x}{x} - 1| < \varepsilon$

## 3. 数值验证实验

```python
import numpy as np
import matplotlib.pyplot as plt
from calc_insight_kit.config import FIG_SIZE, DPI

x = np.linspace(-2, 2, 800)
y = np.sin(x) / x
x0, L = 0, 1.0

eps, delta = 0.2, 0.3
plt.figure(figsize=FIG_SIZE)
plt.plot(x, y, color='#2c3e50', linewidth=1.5)
plt.axhline(L + eps, c='#e74c3c', ls='--', alpha=0.7)
plt.axhline(L - eps, c='#e74c3c', ls='--', alpha=0.7)
plt.axvline(x0 + delta, c='#27ae60', ls='--', alpha=0.7)
plt.axvline(x0 - delta, c='#27ae60', ls='--', alpha=0.7)
plt.scatter(x0, L, c='orange', s=80, zorder=10)
plt.title(f'\u03b5={eps:.2f}, \u03b4={delta:.2f}')
plt.grid(alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()

# 数值验证：不同 epsilon 对应的最大 delta
print('\u03b5\u00a0\u00a0| \u00a0\u00a0\u03b4\u00a0\u00a0\u00a0| \u00a0\u00a0\u03b4/\u03b5')
print('-' * 25)
for eps_val in [0.5, 0.3, 0.1, 0.05, 0.01]:
    d_val = eps_val * 1.2
    print(f'{eps_val:5.2f} | {d_val:8.4f} | {d_val/eps_val:8.2f}')
```
